# Chapitre 1 : Pourquoi NumPy ?

## 1.1 Le problème

Vous connaissez les listes Python. Elles sont très pratiques pour **ranger** des éléments, mais peu pratiques pour **calculer** sur beaucoup de nombres.

Exemple : on a une liste de prix et on veut **doubler chaque prix**.

````python
prix = [10, 20, 30, 40]

print(prix * 2)
````

**Explication :** pour une liste, `* 2` signifie « **répéter** la liste deux fois », et non « multiplier chaque élément ».

Pour doubler les prix, il faut écrire une boucle :

````python
prix_doubles = []
for p in prix:
    prix_doubles.append(p * 2)

print(prix_doubles)
````

Ça fonctionne, mais c'est long pour une opération aussi simple. Et sur des millions de valeurs (ce qui est courant en IA), une boucle Python est **lente**.

## 1.2 La solution : NumPy

**NumPy** (*Numerical Python*) est une bibliothèque spécialisée dans le calcul sur des collections de nombres.

On l'importe toujours sous le nom court `np` (c'est une convention que tout le monde utilise) :

````python
import numpy as np
````

````python
prix_np = np.array([10, 20, 30, 40])

print(prix_np * 2)
````

**Explication :**

1. `np.array([10, 20, 30, 40])` transforme la liste en **tableau NumPy** ;
2. `prix_np * 2` multiplie **chaque élément** par 2.

Nous n'avons pas écrit de boucle : NumPy la fait en interne, beaucoup plus vite.

## 1.3 La vectorisation

On appelle **vectorisation** le fait d'écrire une opération **une seule fois** pour qu'elle s'applique à **tous les éléments** d'un tableau.

| | Avec une liste | Avec NumPy |
|---|---|---|
| Code | boucle `for` + `append` | `prix_np * 2` |
| Lisibilité | on lit la mécanique | on lit la formule |
| Vitesse sur beaucoup de données | lente | rapide |

## 1.4 Qu'est-ce qu'un `ndarray` ?

Le tableau NumPy a un type précis : `ndarray`, pour *n-dimensional array* (« tableau à n dimensions »).

````python
print(type(prix))
print(type(prix_np))
````

## 1.5 Liste ou tableau NumPy : ce qu'il faut retenir

| | Liste Python | Tableau NumPy |
|---|---|---|
| Contenu | types mélangés possibles | **un seul type** pour tous les éléments |
| `* 2` | répète la liste | multiplie chaque élément |
| `+` entre deux collections | colle les deux listes | additionne élément par élément |
| Usage | ranger des objets variés | **calculer** sur des nombres |

> **Pourquoi ne pas toujours utiliser NumPy ?**
> Les listes restent idéales pour ranger des objets variés (textes, dictionnaires…) et ajouter des éléments au fur et à mesure. NumPy est fait pour calculer sur des nombres.

> **Quel rapport avec l'IA ?**
> Un modèle d'IA ne manipule que des nombres : une image, un texte ou un tableau de clients sont d'abord transformés en tableaux de nombres. Et **Pandas est construit sur NumPy** : chaque colonne d'un tableau Pandas est un tableau NumPy.

# Chapitre 2 : Créer des tableaux

## 2.1 À partir d'une liste : `np.array()`

**Pourquoi ?** Pour transformer des données que l'on a déjà (une liste) en tableau NumPy.

````python
notes = np.array([12, 15, 9, 18, 11])

print(notes)
````

**Remarque :** `print` affiche un tableau NumPy **sans virgules**. C'est la façon la plus simple de le distinguer d'une liste.

## 2.2 Le type des éléments : `dtype`

**Pourquoi ?** Un tableau NumPy contient **un seul type** d'éléments. L'attribut `dtype` (*data type*) indique lequel.

````python
print(notes.dtype)
````

`int64` = nombres **entiers**. (Sur certains ordinateurs Windows avec une ancienne version de NumPy, on voit `int32` : pour nous, c'est pareil.)

Si on mélange des entiers et un décimal :

````python
mesures = np.array([1, 2, 3.5])

print(mesures)
print(mesures.dtype)
````

**Explication :** NumPy a converti **tous** les éléments en décimaux (`1.` signifie `1.0`), de type `float64`. Comme il ne peut y avoir qu'un seul type, il choisit celui qui peut représenter toutes les valeurs.

> **`dtype`, c'est une fonction ?**
> Non, c'est un **attribut** : une information stockée dans le tableau. On l'écrit **sans parenthèses** : `notes.dtype`. Même chose pour `shape` et `ndim` plus bas.

## 2.3 Une suite de nombres : `np.arange()`

**Pourquoi ?** Pour générer des valeurs régulières sans les taper.

`np.arange()` fonctionne comme `range()`, que vous connaissez déjà : la **fin est exclue**.

````python
print(np.arange(5))
````

````python
print(np.arange(1, 6))
````

````python
print(np.arange(0, 20, 5))
````

| Appel | Signification | Résultat |
|---|---|---|
| `np.arange(5)` | de 0 à 5 exclu | `[0 1 2 3 4]` |
| `np.arange(1, 6)` | de 1 à 6 exclu | `[1 2 3 4 5]` |
| `np.arange(0, 20, 5)` | de 0 à 20 exclu, de 5 en 5 | `[ 0  5 10 15]` |

## 2.4 Un tableau de zéros ou de uns : `np.zeros()`, `np.ones()`

**Pourquoi ?** Pour préparer un tableau d'une taille donnée, avant de le remplir (par exemple des compteurs qui démarrent à 0).

````python
print(np.zeros(4))
````

````python
print(np.ones(3))
````

Le point après chaque chiffre (`0.`) indique que ce sont des **décimaux** : c'est le choix par défaut de ces deux fonctions.

## 2.5 La forme : `shape` et `ndim`

**Pourquoi ?** Avant de travailler sur des données, on vérifie leur forme. On fera de même avec Pandas.

````python
print(notes)
print("shape :", notes.shape)
print("ndim  :", notes.ndim)
````

- `shape` vaut `(5,)` : **5 éléments** sur une seule dimension. La virgule indique simplement que c'est un tuple d'un seul élément.
- `ndim` vaut `1` : le tableau a **une dimension**. On dit que c'est un tableau **1D** : une simple ligne de valeurs.

## 2.6 Les tableaux à deux dimensions (2D)

Les données réelles ressemblent plutôt à un **tableau Excel**, avec des lignes et des colonnes.

Exemple : les notes de 2 élèves à 3 contrôles.

| | Contrôle 1 | Contrôle 2 | Contrôle 3 |
|---|---|---|---|
| **Léa** | 12 | 15 | 9 |
| **Tom** | 8 | 11 | 10 |

On le crée avec une **liste de listes** : chaque liste intérieure est **une ligne**.

````python
notes_2d = np.array([
    [12, 15, 9],   # Léa
    [8, 11, 10],   # Tom
])

print(notes_2d)
````

````python
print("shape :", notes_2d.shape)
print("ndim  :", notes_2d.ndim)
````

**Lecture :**

- `shape` vaut `(2, 3)` : **2 lignes, 3 colonnes**. L'ordre est **toujours (lignes, colonnes)** ;
- `ndim` vaut `2` : deux dimensions, les lignes et les colonnes.

| | Tableau 1D | Tableau 2D |
|---|---|---|
| Image mentale | une ligne de valeurs | un tableau avec lignes et colonnes |
| Création | `np.array([1, 2, 3])` | `np.array([[1, 2, 3], [4, 5, 6]])` |
| `shape` | `(3,)` | `(2, 3)` |
| `ndim` | 1 | 2 |

### Créer directement un tableau 2D de zéros

`np.zeros` et `np.ones` peuvent aussi créer un tableau 2D : on donne alors la forme `(lignes, colonnes)` entre parenthèses, d'où les **doubles parenthèses**.

````python
print(np.zeros((2, 3)))
````

## 2.7 Changer la forme : `reshape()`

**Pourquoi ?** Pour **réorganiser** des valeurs sans les modifier. Par exemple, ranger 6 valeurs en 2 lignes de 3.

Partons d'un tableau 1D de 6 valeurs :

````python
valeurs = np.arange(6)

print(valeurs)
print("shape :", valeurs.shape)
````

`reshape(2, 3)` range ces 6 valeurs en **2 lignes et 3 colonnes** :

````python
matrice = valeurs.reshape(2, 3)

print(matrice)
print("shape :", matrice.shape)
````

**Explication :** NumPy remplit les cases **ligne par ligne**, dans l'ordre :

```
Avant (1D, shape (6,)) :   [0  1  2  3  4  5]
                            └──┬───┘  └──┬───┘
                           ligne 0    ligne 1

Après (2D, shape (2, 3)) : [[0  1  2]     ← ligne 0
                            [3  4  5]]    ← ligne 1
```

On peut aussi ranger les mêmes valeurs en 3 lignes de 2 :

````python
print(valeurs.reshape(3, 2))
````

### La règle : le nombre d'éléments ne change pas

| Forme demandée | Cases | Possible avec 6 valeurs ? |
|---|---|---|
| `(2, 3)` | 2 × 3 = 6 | oui |
| `(3, 2)` | 3 × 2 = 6 | oui |
| `(4, 2)` | 4 × 2 = 8 | non, il manquerait 2 valeurs |

Essayons quand même `(4, 2)`. On place le code dans `try` / `except` : Python **essaie** d'exécuter le code et, en cas d'erreur, affiche le message au lieu d'arrêter le notebook. (Nous l'utiliserons uniquement pour montrer des erreurs.)

````python
try:
    valeurs.reshape(4, 2)
except ValueError as erreur:
    print("Erreur :", erreur)
````

Le message se lit : « impossible de ranger un tableau de **6** éléments dans la forme **(4, 2)** ».

> **`reshape` modifie-t-il mon tableau ?**
> Non : il renvoie un **nouveau** tableau. `valeurs` garde sa forme d'origine, comme le montre la cellule suivante.

````python
print(valeurs)
````

### Application concrète

Les ventes de 6 mois, rangées en 2 trimestres de 3 mois :

````python
ventes_mois = np.array([100, 120, 90, 130, 110, 150])

ventes_trimestres = ventes_mois.reshape(2, 3)
print(ventes_trimestres)
````

Ligne 0 = premier trimestre, ligne 1 = second trimestre.

# Chapitre 3 : Calculer sans boucle

## 3.1 Opérations avec un nombre

**Pourquoi ?** Pour appliquer le même calcul à toutes les valeurs : ajouter une taxe, convertir une unité, appliquer une remise…

Quand on combine un tableau et **un seul nombre**, l'opération s'applique à **chaque élément**.

````python
prix = np.array([10, 20, 30, 40])

print(prix + 5)
````

**Explication :** 5 a été ajouté à chacun des 4 prix : `10+5`, `20+5`, `30+5`, `40+5`.

Les autres opérations fonctionnent de la même façon :

````python
print(prix - 5)
````

````python
print(prix * 2)
````

````python
print(prix / 2)
````

**Remarque :** la division `/` donne toujours des **décimaux** (`5.`), même entre entiers. C'est le même comportement qu'en Python.

### Application : ajouter la TVA (21 %)

Ajouter 21 % revient à multiplier par 1.21.

````python
prix_ttc = prix * 1.21

print(prix_ttc)
````

## 3.2 Opérations entre deux tableaux

**Pourquoi ?** Pour combiner deux séries de données : additionner les ventes de deux mois, calculer prix × quantité, etc.

Avec deux tableaux de **même taille**, NumPy associe les éléments **à la même position**.

Rappel : avec des listes, `+` colle les deux listes bout à bout.

````python
print([120, 80, 45] + [100, 95, 60])
````

Avec NumPy, `+` additionne **position par position** :

```
  janvier : [120,  80,  45]
+ février : [100,  95,  60]
            ---------------
=           [220, 175, 105]
```

````python
janvier = np.array([120, 80, 45])
fevrier = np.array([100, 95, 60])

print(janvier + fevrier)
````

````python
print(fevrier - janvier)
````

**Lecture :** `fevrier - janvier` donne l'**évolution** de chaque produit entre les deux mois : `-20` = baisse de 20, `+15` = hausse de 15.

### Application : prix × quantités

````python
prix_unitaires = np.array([2, 5, 3])
quantites = np.array([4, 1, 10])

montants = prix_unitaires * quantites
print(montants)
````

**Explication :** `2 × 4 = 8`, `5 × 1 = 5`, `3 × 10 = 30`. Chaque prix est multiplié par **sa** quantité.

## 3.3 Les deux tableaux doivent avoir la même taille

Si les tailles diffèrent, NumPy ne sait pas quel élément associer à quel autre :

````python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30, 40])

try:
    print(a + b)
except ValueError as erreur:
    print("Erreur :", erreur)
````

Le message *operands could not be broadcast together with shapes (3,) (4,)* signifie : « impossible d'associer un tableau de 3 éléments et un tableau de 4 éléments ».

**Réflexe :** quand un calcul échoue, affichez le `shape` des tableaux.

## 3.4 Ça marche aussi en 2D

Un nombre s'applique à **toutes les cases** d'un tableau 2D, exactement comme en 1D. Exemple : un stock en cartons, converti en nombre de pièces (10 pièces par carton).

````python
stock_lundi = np.array([
    [10, 5, 8],
    [3, 7, 2],
])

print(stock_lundi * 10)
````

Chaque case a été multipliée par 10, et le tableau garde sa forme `(2, 3)`.

### Deux tableaux 2D de même forme

Avec deux tableaux 2D de **même forme**, chaque case est associée à la case située au même endroit.

````python
livraison = np.array([
    [4, 6, 1],
    [9, 2, 5],
])

print(stock_lundi + livraison)
````

> **Et si je multiplie un tableau 2D par un tableau 1D ?**
> Dans certains cas, NumPy « étire » automatiquement le petit tableau : c'est le **broadcasting**. Vous l'avez déjà utilisé sans le savoir avec `prix * 2` (le nombre 2 est appliqué partout). Les autres cas ne sont pas au programme du jour : retenez « même forme, ou un seul nombre ».

## Récapitulatif du chapitre 3

| Opération | Exemple | Résultat |
|---|---|---|
| tableau et nombre | `np.array([1, 2, 3]) * 10` | `[10 20 30]` |
| tableau et tableau (même taille) | `np.array([1, 2]) + np.array([10, 20])` | `[11 22]` |
| tailles différentes | `(3,)` + `(4,)` | `ValueError` |

# Chapitre 4 : Sélectionner et filtrer

Sélectionner une partie des données est l'opération la plus courante, en NumPy comme en Pandas.

## 4.1 Accéder à un élément (1D)

Comme pour une liste : des **crochets** et la **position** (l'indice), qui commence à **0**.

| Indice | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| Jour | lun | mar | mer | jeu | ven | sam | dim |
| Température | 12 | 15 | 9 | 21 | 18 | 7 | 14 |

````python
temperatures = np.array([12, 15, 9, 21, 18, 7, 14])

print(temperatures[0])
````

````python
print(temperatures[3])
````

````python
print(temperatures[-1])
````

- `temperatures[0]` → le premier élément (lundi) : 12 ;
- `temperatures[3]` → l'indice 3 (jeudi) : 21 ;
- `temperatures[-1]` → un indice **négatif** compte depuis la fin : `-1` est le dernier (dimanche) : 14.

## 4.2 Extraire une portion : le slicing

**Pourquoi ?** Pour récupérer plusieurs valeurs **consécutives** : les 3 premiers jours, le week-end…

Syntaxe : `tableau[debut:fin]` → de `debut` **inclus** à `fin` **exclue** (comme `range`).

````python
print(temperatures[1:4])
````

Indices 1, 2 et 3 (le 4 est exclu) : mardi, mercredi, jeudi → `[15  9 21]`.

On peut omettre le début ou la fin :

````python
print(temperatures[:3])   # du début jusqu'à l'indice 3 exclu
print(temperatures[5:])   # de l'indice 5 jusqu'à la fin
````

- `[:3]` → les 3 premiers jours (indices 0, 1, 2) ;
- `[5:]` → le week-end (indices 5 et 6).

> **Pourquoi la fin est-elle exclue ?**
> C'est la même règle que `range()` en Python. Avantage : `t[:3]` contient exactement **3** éléments, et `t[:3]` + `t[3:]` redonne tout le tableau sans doublon.

## 4.3 Accéder à un élément (2D)

En 2D, on donne **deux indices** : `tableau[ligne, colonne]`. Toujours la **ligne d'abord**, comme dans `shape`.

Les notes de 3 élèves à 4 contrôles :

```
                 col 0   col 1   col 2   col 3
ligne 0 (Léa)      12      15       9      14
ligne 1 (Tom)       8      11      10      13
ligne 2 (Inès)     17      16      18      15
```

````python
notes_classe = np.array([
    [12, 15, 9, 14],    # Léa
    [8, 11, 10, 13],    # Tom
    [17, 16, 18, 15],   # Inès
])

print(notes_classe[1, 2])
````

Ligne 1 (Tom), colonne 2 (contrôle 3) → **10**.

## 4.4 Sélectionner une ligne ou une colonne

Le symbole `:` seul signifie **« tout »**.

- `notes_classe[1, :]` → ligne 1, **toutes** les colonnes → toutes les notes de Tom ;
- `notes_classe[:, 2]` → **toutes** les lignes, colonne 2 → les notes de tout le monde au contrôle 3.

````python
print(notes_classe[1, :])
````

````python
print(notes_classe[:, 2])
````

**Remarque :** une colonne extraite s'affiche à l'horizontale. C'est normal : c'est maintenant un simple tableau 1D de 3 valeurs.

> **Pourquoi c'est important ?**
> Dans un tableau de données, chaque **colonne** est une information (âge, revenu…). Extraire une colonne est l'opération la plus fréquente. Avec une liste de listes, il faudrait une boucle ; avec NumPy, `[:, 2]` suffit. En Pandas, ce sera `df["nom_colonne"]`.

## 4.5 Le slicing en 2D

On peut découper les lignes **et** les colonnes : `tableau[lignes, colonnes]`.

````python
print(notes_classe[0:2, 0:2])
````

- lignes `0:2` → lignes 0 et 1 (Léa et Tom) ;
- colonnes `0:2` → colonnes 0 et 1 (contrôles 1 et 2) ;
- résultat : un petit tableau de 2 lignes et 2 colonnes.

## 4.6 Les comparaisons produisent des `True` et des `False`

**Pourquoi ?** Pour répondre à la question : « quelles valeurs respectent une condition ? »

Une comparaison entre un tableau et un nombre s'applique, comme les calculs, à **chaque élément**. Le résultat est un tableau de **booléens**.

````python
print(temperatures)
print(temperatures > 15)
````

Pour chaque température, NumPy a répondu à la question « est-elle **supérieure à 15** ? » :

| Jour | lun | mar | mer | jeu | ven | sam | dim |
|---|---|---|---|---|---|---|---|
| Température | 12 | 15 | 9 | 21 | 18 | 7 | 14 |
| `> 15` ? | False | False | False | **True** | **True** | False | False |

(15 n'est **pas** strictement supérieur à 15, d'où `False` pour mardi.)

Opérateurs disponibles : `>`, `>=`, `<`, `<=`, `==` (égal), `!=` (différent).

## 4.7 Le masque booléen : sélectionner avec des `True` / `False`

Un tableau de booléens utilisé pour sélectionner s'appelle un **masque**.

Placé entre crochets, le masque **garde uniquement les valeurs aux positions `True`**.

````python
masque = temperatures > 15

print("Températures :", temperatures)
print("Masque       :", masque)
print("Sélection    :", temperatures[masque])
````

**Explication, en deux étapes :**

1. `temperatures > 15` produit le masque `[False False False True True False False]` ;
2. `temperatures[masque]` garde les valeurs là où le masque vaut `True` : positions 3 et 4 → `[21 18]`.

En pratique, on écrit les deux étapes en une seule ligne :

````python
print(temperatures[temperatures < 10])
````

**Comment le lire :** « les températures, **là où** les températures sont inférieures à 10 ».

### Application : filtrer un tableau grâce à un autre

On peut calculer la condition sur un tableau et l'appliquer à un **autre tableau de même taille**.

Question : **quels jours** a-t-il fait plus de 15 degrés ?

````python
jours = np.array(["lun", "mar", "mer", "jeu", "ven", "sam", "dim"])

print(jours[temperatures > 15])
````

Le masque est calculé sur `temperatures`, puis appliqué à `jours`. Cela fonctionne parce que la position *i* désigne le **même jour** dans les deux tableaux.

> **Comment compter les valeurs qui respectent la condition ?**
> Nous le verrons au chapitre 5 avec `np.sum` : pour NumPy, `True` compte comme 1 et `False` comme 0.

## 4.8 Combiner deux conditions : `&` (ET) et `|` (OU)

| Opérateur | Signification | Vrai si… |
|---|---|---|
| `&` | ET | les **deux** conditions sont vraies |
| `\|` | OU | **au moins une** condition est vraie |

**Règle :** chaque condition doit être entre **parenthèses**.

````python
# Entre 10 et 20 degrés : supérieur ou égal à 10 ET inférieur ou égal à 20
douces = (temperatures >= 10) & (temperatures <= 20)

print(douces)
print(temperatures[douces])
````

````python
# Températures extrêmes : moins de 10 OU plus de 20
extremes = (temperatures < 10) | (temperatures > 20)

print(jours[extremes])
````

> **Pourquoi pas `and` et `or` comme d'habitude ?**
> `and` et `or` attendent **une seule** valeur `True` ou `False`. Ici, on compare **7 valeurs** d'un coup : Python ne sait pas laquelle prendre et lève une erreur (*The truth value of an array … is ambiguous*). `&` et `|` combinent les booléens **position par position**.

> **Pourquoi les parenthèses ?**
> Python calcule `&` **avant** `>=`. Sans parenthèses, `temperatures >= 10 & temperatures <= 20` serait lu de travers et produirait une erreur.

### Application : filtrer les lignes d'un tableau 2D

Chaque ligne est un client, chaque colonne une information : âge, puis revenu annuel (en milliers d'euros).

````python
clients = np.array([
    [25, 32],
    [41, 58],
    [19, 15],
    [35, 71],
])

ages = clients[:, 0]
print("Âges :", ages)
print("Masque :", ages < 30)
print(clients[ages < 30])
````

**Explication :**

1. `clients[:, 0]` extrait la colonne des âges ;
2. `ages < 30` donne un masque avec **une valeur par client** ;
3. `clients[ages < 30]` garde les **lignes complètes** des clients de moins de 30 ans.

> En Pandas, on écrira `clients[clients["age"] < 30]`. Le principe est le même.

# Chapitre 5 : Statistiques et `axis`

Une **agrégation** résume plusieurs valeurs en une seule : un total, une moyenne, un minimum, un maximum. C'est souvent la première étape pour comprendre des données.

## 5.1 Les quatre fonctions de base

Les ventes d'une boutique sur 5 jours :

````python
ventes = np.array([120, 90, 150, 80, 110])

print(np.sum(ventes))
````

`np.sum` additionne tous les éléments : 120 + 90 + 150 + 80 + 110 = **550**.

````python
print(np.mean(ventes))
````

`np.mean` calcule la **moyenne** : 550 ÷ 5 = **110**.

````python
print(np.min(ventes))
print(np.max(ventes))
````

`np.min` et `np.max` donnent la plus petite (**80**) et la plus grande (**150**) valeur.

> **J'ai vu `ventes.sum()` sur Internet. C'est différent ?**
> Non, c'est la même chose écrite autrement : `np.sum(ventes)` et `ventes.sum()` donnent le même résultat. Dans ce cours, on utilise la forme `np.…`.

## 5.2 Statistiques et filtrage

On combine le chapitre 4 et le chapitre 5 : d'abord **sélectionner**, ensuite **résumer**.

**Compter** les jours à plus de 100 ventes : on additionne le masque, car `True` vaut 1 et `False` vaut 0.

````python
print(ventes > 100)
print(np.sum(ventes > 100))
````

Le masque contient 3 fois `True` → 1 + 0 + 1 + 0 + 1 = **3 jours**.

**Moyenne d'une sélection :** la moyenne des « bons jours » uniquement.

````python
bons_jours = ventes[ventes > 100]

print(bons_jours)
print(np.mean(bons_jours))
````

(120 + 150 + 110) ÷ 3 ≈ **126.67**.

## 5.3 Les tableaux 2D et le paramètre `axis`

Sur un tableau 2D, on veut souvent un résultat **par ligne** ou **par colonne**, et pas seulement un total général. C'est le rôle du paramètre `axis`.

Exemple : 2 magasins (lignes) et 3 jours (colonnes).

| | Lundi | Mardi | Mercredi |
|---|---|---|---|
| **Magasin A** | 10 | 20 | 30 |
| **Magasin B** | 40 | 50 | 60 |

````python
ventes_magasins = np.array([
    [10, 20, 30],   # magasin A
    [40, 50, 60],   # magasin B
])

print(ventes_magasins.shape)
````

### Sans `axis` : un seul nombre pour tout le tableau

````python
print(np.sum(ventes_magasins))
````

10 + 20 + 30 + 40 + 50 + 60 = **210** : le total des deux magasins sur les trois jours.

### `axis=0` : un résultat **par colonne**

On additionne les valeurs **de chaque colonne** :

| | Lundi | Mardi | Mercredi |
|---|---|---|---|
| Magasin A | 10 | 20 | 30 |
| Magasin B | 40 | 50 | 60 |
| **`axis=0` →** | **10 + 40 = 50** | **20 + 50 = 70** | **30 + 60 = 90** |

````python
print(np.sum(ventes_magasins, axis=0))
````

Résultat : `[50 70 90]` → **3 valeurs, une par colonne** = le total **par jour**, tous magasins confondus.

### `axis=1` : un résultat **par ligne**

On additionne les valeurs **de chaque ligne** :

| | Lundi | Mardi | Mercredi | **`axis=1` →** |
|---|---|---|---|---|
| Magasin A | 10 | 20 | 30 | **10 + 20 + 30 = 60** |
| Magasin B | 40 | 50 | 60 | **40 + 50 + 60 = 150** |

````python
print(np.sum(ventes_magasins, axis=1))
````

Résultat : `[ 60 150]` → **2 valeurs, une par ligne** = le total **par magasin**.

### Comment ne pas se tromper

| Je veux… | J'écris | Nombre de résultats (pour une forme `(2, 3)`) |
|---|---|---|
| un seul total | `np.sum(t)` | 1 |
| un résultat **par colonne** | `np.sum(t, axis=0)` | 3 (le nombre de colonnes) |
| un résultat **par ligne** | `np.sum(t, axis=1)` | 2 (le nombre de lignes) |

**Méthode en 2 questions :**

1. *« Je veux un résultat par quoi ? »* → par colonne : `axis=0` ; par ligne : `axis=1`.
2. *Vérification :* je compte les valeurs obtenues. 3 valeurs pour 3 colonnes ? C'est bien un résultat par colonne.

> **Pourquoi `axis=0` donne-t-il un résultat par colonne ? Ça paraît à l'envers.**
> Dans `shape = (2, 3)`, la position 0 correspond aux **lignes** (2) et la position 1 aux **colonnes** (3). `axis=0` signifie « on **combine les lignes** entre elles » : les 2 lignes fusionnent, il reste une valeur par colonne. Si cette explication ne parle pas, retenez simplement le tableau ci-dessus et vérifiez toujours le nombre de résultats.

### `axis` fonctionne avec toutes les fonctions

````python
print(np.mean(ventes_magasins, axis=1))
````

Moyenne **par magasin** (par ligne) : A = 60 ÷ 3 = **20**, B = 150 ÷ 3 = **50**.

````python
print(np.max(ventes_magasins, axis=0))
````

Meilleure vente **de chaque jour** (par colonne) : le maximum de chaque colonne → `[40 50 60]`.

> En Pandas, `df.mean()` calcule par défaut **une moyenne par colonne** : c'est l'équivalent de `axis=0`.

## Récapitulatif du chapitre 5

| Fonction | Rôle |
|---|---|
| `np.sum(t)` | somme |
| `np.mean(t)` | moyenne |
| `np.min(t)` / `np.max(t)` | minimum / maximum |
| `np.sum(t > 10)` | compter les valeurs qui respectent une condition |
| `axis=0` | un résultat **par colonne** |
| `axis=1` | un résultat **par ligne** |

# Chapitre 6 : Valeurs manquantes

## 6.1 Pourquoi ?

Dans les vraies données, il manque souvent des valeurs : capteur en panne, formulaire incomplet, ticket perdu… Il faut savoir les **repérer**, sinon les calculs ne donnent plus de résultat.

## 6.2 `np.nan` : la valeur manquante

NumPy représente une valeur manquante par `np.nan` (*Not a Number*, « pas un nombre »).

Exemple : les températures d'une semaine, avec deux relevés manquants (mercredi et samedi).

````python
releves = np.array([12, 15, np.nan, 21, 18, np.nan, 14])

print(releves)
````

Les valeurs manquantes s'affichent `nan`. Les autres nombres sont devenus des décimaux (`12.`) : `np.nan` est un décimal, et un tableau n'a qu'un seul type.

> En Pandas, les valeurs manquantes s'afficheront `NaN`. C'est la même chose.

## 6.3 Le problème : les calculs donnent `nan`

````python
print(np.mean(releves))
````

**Explication :** une seule valeur inconnue rend le résultat inconnu. NumPy refuse de « deviner » et renvoie `nan`. Même chose avec `np.sum`, `np.min` ou `np.max`.

## 6.4 Repérer les valeurs manquantes : `np.isnan()`

**Piège :** on ne peut **pas** tester `== np.nan`. Par définition, `nan` n'est égal à rien, **pas même à lui-même**.

````python
print(np.nan == np.nan)
````

Il faut utiliser la fonction **`np.isnan()`**. Elle renvoie un **masque** : `True` là où la valeur manque.

````python
print(releves)
print(np.isnan(releves))
````

Avec ce masque, on réutilise ce que l'on connaît déjà :

````python
jours = np.array(["lun", "mar", "mer", "jeu", "ven", "sam", "dim"])

print("Nombre de valeurs manquantes :", np.sum(np.isnan(releves)))
print("Jours sans relevé :", jours[np.isnan(releves)])
````

## 6.5 Calculer malgré les valeurs manquantes : `np.nanmean()`

`np.nanmean()` calcule la moyenne **en ignorant** les valeurs manquantes.

````python
print(np.nanmean(releves))
````

**Explication :** la moyenne est calculée sur les **5 valeurs présentes** :
(12 + 15 + 21 + 18 + 14) ÷ 5 = 80 ÷ 5 = **16**.

Elle accepte aussi `axis`. Deux villes, quatre jours, un relevé manquant pour la ville A :

````python
temperatures_villes = np.array([
    [12, 15, np.nan, 21],   # ville A
    [8, 10, 11, 9],         # ville B
])

print(np.nanmean(temperatures_villes, axis=1))
````

Moyenne **par ville** (par ligne) : A = (12 + 15 + 21) ÷ 3 = **16**, B = 38 ÷ 4 = **9.5**.

> **Il existe aussi `np.nansum`, `np.nanmax`… ?**
> Oui, chaque fonction a sa version `nan…` qui ignore les valeurs manquantes. Aujourd'hui, `np.nanmean` suffit.

## 6.6 Remplacer une valeur manquante

Autre stratégie : **remplacer** chaque `nan` par une valeur choisie. On combine le masque et l'affectation :

```python
tableau[masque] = nouvelle_valeur
```

On le fait sur une **copie**, pour garder les données d'origine : `.copy()` crée un vrai double du tableau.

````python
releves_complets = releves.copy()
releves_complets[np.isnan(releves_complets)] = 0

print(releves_complets)
````

**Lecture :** « dans `releves_complets`, **là où** la valeur manque, mettre 0 ».

Mais attention : 0 °C est une **vraie** température. Ce remplacement fausse la moyenne :

````python
print(np.mean(releves_complets))
````

80 ÷ 7 ≈ **11.43**, au lieu de 16. Ici, `np.nanmean` donne le résultat le plus juste.

**Il n'y a pas de bonne réponse universelle :** remplacer par 0 a du sens si l'absence veut vraiment dire « zéro » (par exemple : aucun achat). Sinon, mieux vaut ignorer les valeurs manquantes.

> En Pandas : `isna()` pour repérer, `fillna()` pour remplacer, et `mean()` ignore les `NaN` par défaut.

|---|---|

| `np.nanmean(t)` | moyenne en ignorant les valeurs manquantes |
| `copie = t.copy()` puis `copie[np.isnan(copie)] = 0` | remplacer les valeurs manquantes |
